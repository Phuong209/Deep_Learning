```python
import torch

A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A:")
print(A)
print("B:")
print(B)

print("\nA + B:")
print(A + B)
print("\nA * B (element-wise):")
print(A * B)
print("\nA / B:")
print(A / B)
print("\nMean theo cột (dim=0):")
print(torch.mean(A, dim=0))
print("\nStd theo cột (dim=0):")
print(torch.std(A, dim=0))
```

```python
import torch

x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

print("x =", x)
print("y =", y)
print("dy/dx (autograd) =", x.grad)

manual_grad = 4 * 2 + 5  # y' = 4x + 5
print("dy/dx (tính thủ công tại x=2) =", manual_grad)
```

```python
import numpy as np
import torch

polynomial = np.poly1d([2, -3, 5])   # y = 2x^2 - 3x + 5
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(3000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients (kỳ vọng gần [2, -3, 5]):")
print(w)
```
